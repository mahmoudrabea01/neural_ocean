```python
!pip install roboflow

from roboflow import Roboflow
rf = Roboflow(api_key="You_API")
project = rf.workspace("neural-ocean").project("neural_ocean")
version = project.version(3)
dataset = version.download("yolo26")
```

```python
from ultralytics import YOLO

model = YOLO("yolo26s.pt")
```

```python
results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=60,
    imgsz=640,
    batch=16,
    device=0,
    workers=8,
    project="/content/runs/neural_ocean",
    name="yolo26s",
    patience=10,
    exist_ok=True
)
```

```python
from pathlib import Path

RESULTS_DIR = Path(results.save_dir)

print("Results directory:", RESULTS_DIR)
print("\nGenerated files:")

for file in sorted(RESULTS_DIR.iterdir()):
    print(file.name)
```

```python
WEIGHTS_DIR = RESULTS_DIR / "weights"

for file in WEIGHTS_DIR.iterdir():
    print(f"{file.name} | {file.stat().st_size / (1024**2):.2f} MB")
```

```python
from ultralytics import YOLO

best_model = YOLO(WEIGHTS_DIR / "best.pt")

test_metrics = best_model.val(
    data=f"{dataset.location}/data.yaml",
    split="test",
    imgsz=640,
    batch=16,
    device=0
)
```
